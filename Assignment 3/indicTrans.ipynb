## Installing required Libraries

```python
import torch
from transformers import AutoModelForSeq2SeqLM
from IndicTransTokenizer import IndicProcessor, IndicTransTokenizer
```

## Loading Tokenizer and IndicTrans Model

```python
tok_en_indic = IndicTransTokenizer(direction="en-indic")
ip = IndicProcessor(inference=True)
eng_lang_model = AutoModelForSeq2SeqLM.from_pretrained("ai4bharat/indictrans2-en-indic-dist-200M", trust_remote_code=True)
```

```python
tok_indic_en = IndicTransTokenizer(direction="indic-en")
ip = IndicProcessor(inference=True)
indic_model = AutoModelForSeq2SeqLM.from_pretrained("ai4bharat/indictrans2-indic-en-dist-200M", trust_remote_code=True)
```

```python
tok_indic_en_indic = IndicTransTokenizer(direction="indic-indic")
ip = IndicProcessor(inference=True)
indic_indic_model = AutoModelForSeq2SeqLM.from_pretrained("ai4bharat/indictrans2-indic-indic-dist-320M", trust_remote_code=True)
```

### Importing English File

```python
with open("/kaggle/input/dataset3/test.en","r",encoding="utf-8") as file:
    sen_in_english=file.readlines()

sen_in_english=[sentence for sentence in sen_in_english]
```

### Importing Hindi File

```python
with open("/kaggle/input/dataset3/test.hi","r",encoding="utf-8") as file:
    sen_in_hindi=file.readlines()

sen_in_hindi=[sentence for sentence in sen_in_hindi]
```

### Importing Gujrati File

```python
with open("/kaggle/input/dataset3/test.gu","r",encoding="utf-8") as file:
    sen_in_gujarati=file.readlines()

sen_in_gujarati=[sentence for sentence in sen_in_gujarati]
```

### Randomly Selecting a Thousand Sentences from Hindi, English, and Gujarati Data and using seed to take care of same random sentences for all three languages 

```python
import random
random.seed(1)
index_of_sampled_sentences = random.sample(range(len(sen_in_english)), 1000)
english_sentences_sampled = [sen_in_english[i] for i in index_of_sampled_sentences]
hindi_sentences_sampled = [sen_in_hindi[i] for i in index_of_sampled_sentences]
gujarati_sentences_sampled = [sen_in_gujarati[i] for i in index_of_sampled_sentences]
```

# Translating English to Hindi

```python
english_to_hindi_translation=[]
for i in range(0,20):
    batch = ip.preprocess_batch(english_sentences_sampled[i*50:(i*50)+50], src_lang="eng_Latn", tgt_lang="hin_Deva")
    batch = tok_en_indic(batch, src=True, return_tensors="pt")
    print(i)

    with torch.inference_mode():
        outputs = eng_lang_model.generate(**batch, num_beams=5, num_return_sequences=1, max_length=256)

    outputs = tok_en_indic.batch_decode(outputs, src=False)
    english_to_hindi_translation.extend(ip.postprocess_batch(outputs, lang="hin_Deva"))
```

#### For rerun I'm saving the translated file locally

```python
hindi_to_english_translation=[]
for i in range(0,20):
    batch = ip.preprocess_batch(hindi_sentences_sampled[i*50:(i*50)+50], src_lang="hin_Deva", tgt_lang="eng_Latn")
    batch = tok_indic_en(batch, src=True, return_tensors="pt")

    with torch.inference_mode():
        outputs = indic_model.generate(**batch, num_beams=5, num_return_sequences=1, max_length=256)

    outputs = tok_indic_en.batch_decode(outputs, src=False)
    hindi_to_english_translation.extend(ip.postprocess_batch(outputs, lang="eng_Latn"))
```

# Evaluation of models

### Calculation of BLEU Score

```python
from nltk.translate.bleu_score import corpus_bleu

def calculate_bleu(reference_texts, translated_texts):
    """
    Args:
        reference_texts (list): A list of correct reference translations.
        translated_texts (list): A list of machine-generated translations to evaluate.
        
    Returns:
        float: The BLEU score for the translations.
    """
    # Format references as a list of lists for BLEU calculation
    formatted_references = [[ref] for ref in reference_texts]
    # Calculate and return the BLEU score
    bleu_score = corpus_bleu(formatted_references, translated_texts)
    return bleu_score

def print_bleu_score(description, reference_texts, translated_texts):
    """
    
    Args:
        description (str): Description of the translation (e.g., "English to Hindi").
        reference_texts (list): Reference texts for BLEU score calculation.
        translated_texts (list): Translated texts to evaluate.
    """
    bleu_score = calculate_bleu(reference_texts, translated_texts)
    print(f"BLEU Score for {description}: {bleu_score:.4f}")


# Printing BLEU scores with formatted output
print_bleu_score("English to Hindi", hindi_sentences_sampled, english_to_hindi_translation)
print_bleu_score("Hindi to English", english_sentences_sampled, hindi_to_english_translation)
print_bleu_score("Hindi to Gujarati", gujarati_sentences_sampled, hindi_to_gujarati_translation)
print_bleu_score("Gujarati to Hindi", hindi_sentences_sampled, gujarati_to_hindi_translation)
```

```python
from nltk.translate.bleu_score import sentence_bleu

# Function to calculate sentence-level BLEU scores and return the average
def calculate_average_sentence_bleu(references, translations):
    score=0
    for ref, trans in zip(references, translations):
        score += sentence_bleu([ref], trans)
    # Calculate the average of the sentence BLEU scores
    average_score = score / 1000 #if scores else 0
    return average_score

# Function to print the average sentence-level BLEU score
def print_average_sentence_bleu(model_name, references, translations):
    average_score = calculate_average_sentence_bleu(references, translations)
    print(f"\nAverage Sentence-level BLEU Score for {model_name}: {average_score:.4f}")

# Calculating Average Sentence-level BLEU Score for each translation model
print_average_sentence_bleu("English to Hindi", hindi_sentences_sampled, english_to_hindi_translation)
print_average_sentence_bleu("Hindi to English", english_sentences_sampled, hindi_to_english_translation)
print_average_sentence_bleu("Hindi to Gujarati", gujarati_sentences_sampled, hindi_to_gujarati_translation)
print_average_sentence_bleu("Gujarati to Hindi", hindi_sentences_sampled, gujarati_to_hindi_translation)
```

```python
import numpy as np
from rouge import Rouge

# Initialize the Rouge scorer
rouge = Rouge()

def average_rouge_scores(translated_texts, reference_texts):
    """

    Args:
        translated_texts (list of str): The translated sentences.
        reference_texts (list of str): The ground truth reference sentences.

    Returns:
        dict: A dictionary containing averaged scores for ROUGE-1, ROUGE-2, and ROUGE-L.
    """
    scores = rouge.get_scores(translated_texts, reference_texts, avg=True)  # Use avg=True to get the average score
    return scores
def print_rouge_scores(model_name, scores):
    """

    Args:
    model_name (str): Description of the model translation direction.
    scores (dict): A dictionary containing averaged ROUGE scores.
    """
    print(f"\nROUGE Scores for {model_name} Translation:")
    print(f"{'Metric':<8} | {'Precision':<10} | {'Recall':<10} | {'F1-score':<10}")
    print('-' * 42)
    for key in ['rouge-1', 'rouge-2', 'rouge-l']:
        metric_name = key.upper().replace('ROUGE-', 'ROUGE ')
        precision = scores[key]['p']
        recall = scores[key]['r']
        f1_score = scores[key]['f']
        print(f"{metric_name:<8} | {precision:<10.4f} | {recall:<10.4f} | {f1_score:<10.4f}")

# Example ROUGE scores printing for each model translation
rouge_score_eng_to_hin = average_rouge_scores(english_to_hindi_translation, hindi_sentences_sampled)
print_rouge_scores("English to Hindi", rouge_score_eng_to_hin)

rouge_score_hin_to_eng = average_rouge_scores(hindi_to_english_translation, english_sentences_sampled)
print_rouge_scores("Hindi to English", rouge_score_hin_to_eng)

rouge_score_hin_to_guj = average_rouge_scores(hindi_to_gujarati_translation, gujarati_sentences_sampled)
print_rouge_scores("Hindi to Gujarati", rouge_score_hin_to_guj)

rouge_score_guj_to_hin = average_rouge_scores(gujarati_to_hindi_translation, hindi_sentences_sampled)
print_rouge_scores("Gujarati to Hindi", rouge_score_guj_to_hin)
```
