```python
from transformers import AutoModelForSeq2SeqLM, AutoTokenizer
import random


tokenizer = AutoTokenizer.from_pretrained("facebook/nllb-200-distilled-600M")
model = AutoModelForSeq2SeqLM.from_pretrained("facebook/nllb-200-distilled-600M")
```

# Importing English File

```python
with open("/kaggle/input/dataset/test.en", 'r') as file:
    
    sentences_eng = file.readlines()
    
sentences_eng = [sentence.strip() for sentence in sentences_eng]
```

# Importing Hindi File

```python
with open("/kaggle/input/dataset/test.hi", 'r') as file:
    
    sentences_hi = file.readlines()
    
sentences_hi = [sentence.strip() for sentence in sentences_hi]
```

# Importing Gujrati File

```python
with open("/kaggle/input/dataset/test.gu", 'r') as file:
    
    sentences_gu = file.readlines()
    
sentences_gu = [sentence.strip() for sentence in sentences_gu]
```

# Sampling random thousand sentences from Hindi, English and Gujrati Data 

```python
sample_indices = random.sample(range(len(sentences_eng)), 1000)
sample_english = [sentences_eng[i] for i in sample_indices]
sample_hindi = [sentences_hi[i] for i in sample_indices]
sample_gujrati = [sentences_gu[i] for i in sample_indices]
```

```python
# sample_english
```

```python
# sample_hindi
```

```python
# sample_gujrati
```

# Translating English to Hindi

```python
articles = sample_english
inputs = tokenizer(articles, return_tensors="pt", padding = "longest")

translated_tokens = model.generate(
    **inputs, forced_bos_token_id=tokenizer.lang_code_to_id["hin_Deva"], max_length= 32
)
translated_eng_to_hin = tokenizer.batch_decode(translated_tokens, skip_special_tokens=True)
```

```python
# translated_eng_to_hin
```

# Translating Hindi to English

```python
articles = sample_hindi
inputs = tokenizer(articles, return_tensors="pt", padding = "longest")

translated_tokens = model.generate(
    **inputs, forced_bos_token_id=tokenizer.lang_code_to_id["eng_Latn"], max_length= 32
)
translated_hin_to_eng = tokenizer.batch_decode(translated_tokens, skip_special_tokens=True)
```

```python
# translated_hin_to_eng
```

# Translating Hindi to Gujrati

```python
articles = sample_hindi
inputs = tokenizer(articles, return_tensors="pt", padding = "longest")

translated_tokens = model.generate(
    **inputs, forced_bos_token_id=tokenizer.lang_code_to_id["guj_Gujr"], max_length= 32
)
translated_hin_to_guj = tokenizer.batch_decode(translated_tokens, skip_special_tokens=True)
```

```python
# translated_hin_to_guj
```

# Translating Gujrati to Hindi

```python
articles = sample_gujrati
inputs = tokenizer(articles, return_tensors="pt", padding = "longest")

translated_tokens = model.generate(
    **inputs, forced_bos_token_id=tokenizer.lang_code_to_id["hin_Deva"], max_length= 32
)
translated_guj_to_hin = tokenizer.batch_decode(translated_tokens, skip_special_tokens=True)
```

```python
# translated_guj_to_hin
```

# Evaluation of models

### Calculation of BLEU Score

```python
from nltk.translate.bleu_score import corpus_bleu
```

#### BLEU Score of translation Eng to Hin

```python
bleu_scores_eng_to_hin = corpus_bleu([[ref] for ref in sample_hindi], translated_eng_to_hin)
print("BLEU Score of Translating English to Hindi is :" , bleu_scores_eng_to_hin)
```

#### BLEU Score of translation Hin to Eng

```python
bleu_scores_hin_to_eng = corpus_bleu([[ref] for ref in sample_english], translated_hin_to_eng)
print("BLEU Score of Translating Hindi to Eng is :" , bleu_scores_hin_to_eng)
```

#### BLEU Score of translation Hin to Guj

```python
bleu_scores_hin_to_guj = corpus_bleu([[ref] for ref in sample_gujrati], translated_hin_to_guj)
print("BLEU Score of Translating Hindi to Gujrati is :" , bleu_scores_hin_to_guj)
```

#### BLEU Score of translation Guj to Hin

```python
bleu_scores_guj_to_hin = corpus_bleu([[ref] for ref in sample_hindi], translated_guj_to_hin)
print("BLEU Score of Translating Gujrati to Hindi is :" , bleu_scores_guj_to_hin)
```

### Calculation of ROGUE Score

```python
import numpy as np
from rouge import Rouge

# Initialize the Rouge scorer
rouge = Rouge()

# Function to calculate average ROUGE scores
def average_rouge_scores(translated_texts, reference_texts):
    scores = rouge.get_scores(translated_texts, reference_texts, avg=True)  # Use avg=True to get the average score
    return scores

# Function to print scores in a formatted way
def print_rouge_scores(model_name, scores):
    print(f"\nAverage ROUGE scores for {model_name}:")
    print(f"  ROUGE-1: Precision: {scores['rouge-1']['p']:.4f}, Recall: {scores['rouge-1']['r']:.4f}, F1-score: {scores['rouge-1']['f']:.4f}")
    print(f"  ROUGE-2: Precision: {scores['rouge-2']['p']:.4f}, Recall: {scores['rouge-2']['r']:.4f}, F1-score: {scores['rouge-2']['f']:.4f}")
    print(f"  ROUGE-L: Precision: {scores['rouge-l']['p']:.4f}, Recall: {scores['rouge-l']['r']:.4f}, F1-score: {scores['rouge-l']['f']:.4f}")

# Calculating average ROUGE scores for each translation model
rouge_score_eng_to_hin = average_rouge_scores(translated_eng_to_hin, sample_hindi)
print_rouge_scores("English to Hindi", rouge_score_eng_to_hin)

rouge_score_hin_to_eng = average_rouge_scores(translated_hin_to_eng, sample_english)
print_rouge_scores("Hindi to English", rouge_score_hin_to_eng)

rouge_score_hin_to_guj = average_rouge_scores(translated_hin_to_guj, sample_gujrati)
print_rouge_scores("Hindi to Gujarati", rouge_score_hin_to_guj)

rouge_score_guj_to_hin = average_rouge_scores(translated_guj_to_hin, sample_hindi)
print_rouge_scores("Gujarati to Hindi", rouge_score_guj_to_hin)
```
