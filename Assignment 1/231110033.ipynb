```python
#BPE tokenizer
import re
from collections import Counter, defaultdict

def construct_vocabulary(term):
    x = len(term)
    list_of_tokens = []

    for i in range(x):
 # Separate each char in word by space and add mark end of token
        m = term[i]
        for word in m.split():
            s = " ".join(word) + " </w>" 
            list_of_tokens.append(s)
    
    # Count frequency of tokens in corpus
    vocab = Counter(list_of_tokens)  

    return vocab

def find_pairs(vocab):
    
    pairs_of_symbols = defaultdict(int)
    for word, frequency in vocab.items():
        symbols = word.split()

        # Counting up occurrences of pairs
        for i in range(len(symbols) - 1):
            pairs_of_symbols[symbols[i], symbols[i + 1]] += frequency

    return pairs_of_symbols

def merge_vocab(pair, v_in):
    
    v_out = {}
    bigram = re.escape(' '.join(pair))
    p = re.compile(r'(?<!\S)' + bigram + r'(?!\S)')
    
    for word in v_in:
        # replace most frequent pair in all vocabulary
        w_out = p.sub(''.join(pair), word)
        v_out[w_out] = v_in[word]

    return v_out
```

```python
#Implementing BPE for Vocabulary size 1K
    
vocab = construct_vocabulary(term)  # Step 1
new_word1 = []
num_merges = 1000 # Hyperparameter
for i in range(num_merges):

    pairs = find_pairs(vocab)  # Step 2

    if not pairs:
        break

    # step 3
    best = max(pairs, key=pairs.get)
    new_word1.append(best[0]+best[1])
    #print(best)
    vocab = merge_vocab(best, vocab)
    
new_word1 =' '.join(new_word1).split()
bpe_1k = []
for item in new_word1:
    x = item.replace("</w>","")
    bpe_1k.append(x)
print(bpe_1k)
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score

def calcu_Pre_rec_F(truevalue, predicted, model):
    # Example ground truth and predicted token lists
    ground_truth_tokens = truevalue
    predicted_tokens = predicted

    # Convert lists to sets for easier comparison
    ground_truth_set = set(ground_truth_tokens)
    predicted_set = set(predicted_tokens)

    # Calculate true positives, false positives, and false negatives
    true_positives = ground_truth_set.intersection(predicted_set)
    false_positives = predicted_set - ground_truth_set
    false_negatives = ground_truth_set - predicted_set

    # Calculate precision, recall, and F-score
    precision = len(true_positives) / (len(true_positives) + len(false_positives)) if len(true_positives) > 0 else 0
    recall = len(true_positives) / (len(true_positives) + len(false_negatives)) if len(true_positives) > 0 else 0
    f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0

    # Print the results
    print("The tokenizer used is",model)
    print("The value of precision is:", precision)
    print("The value of recall is:", recall)
    print("The value of f-score is:", f1)
```
